## Data Structures

```python
# Create list of sales values: 100, 250, 80
sales = [100,250,80]

# Create list of tax values: 5, 12.5, 4
tax = [5,12.5,4]
```

```python
new_sale = 40
new_tax = 2

# Append the additional sales and tax values to the lists
sales.append(new_sale)
tax.append(new_tax)

# Print the updated lists
print(sales)
print(tax)
```

## Operators & Functions

```python
# Use the current year and prior year variables to answer the following questions:
CY_taxRate = 0.10
PY_taxRate = 0.12

CY_customerOrders = 500
PY_customerOrders = 300
```

```python
# Is current year tax rate lower than prior year?
CY_taxRate < PY_taxRate
```

```python
# What is the YoY change in customer orders?
CY_customerOrders - PY_customerOrders
```

```python
# Assuming each order cost $30, what tax was made on customer orders in the prior year?
30 * PY_taxRate * PY_customerOrders
```

```python
# What is the lowest tax payment?
min(tax)
```

## Conditional Statements & For Loops

```python
CY_customerOrders = 500
PY_customerOrders = 300
```

```python
# Create a conditional statement that prints "Orders Increased" if current year orders are higher than prior year orders
if CY_customerOrders > PY_customerOrders:
  print("Orders Increased")
```

```python
sales = [100,250,80]
```

```python
# Loop through the values of the sales list to print:
# "High Sales" when amount is greater than 200
# "Med Sales" when amount is greater than 100
# "Low Sales" otherwise

for i in sales:
  if i > 200:
    print("High Sale")
  elif i > 100:
    print("Med Sale")
  else:
    print("Low Sale")
```

```python
sales_dict = {
    "TXN001": 100,
    "TXN002": 250,
    "TXN003":  80
}

tax_dict = {
    "TXN001":   5,
    "TXN002": 12.5,
    "TXN003":   4
}
```

```python
# Loop through the values of sales_dict and tax_dict and subtract tax from sales amount for each txn
for key, value in sales_dict.items():
  print(key, value - tax_dict[key])
```
